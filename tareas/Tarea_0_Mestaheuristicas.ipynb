### Fenómeno Natural: Simbiosis Depredador-Rémora

**Descripción Biológica:**
Las rémoras (*Remora remora*) son peces que practican una estrategia de supervivencia basada en **simbiosis comensalista**:
- Se adhieren a depredadores grandes (tiburones, mantas rayas) mediante un disco adhesivo
- Se alimentan de restos de presas que el depredador no consume
- Reducen drasticamente su energía gastada en búsqueda, acecho y captura
- Ganancia mutua: el depredador recibe limpieza de parasitosis

**Principio Fundamental:**
Explotar estrategias exitosas ya descubiertas (el depredador) mientras exploran **variaciones locales** alrededor de esas buenas soluciones.

### Metáfora vs Modelo

**Metáfora (Lenguaje Natural):**
- "Depredador" = agente explorador con solución buena
- "Rémoras" = agentes explotadores que rodean la solución
- "Alimento restante" = mejora marginal alrededor de la solución conocida
- "Adhesión" = cercanía en espacio de parámetros

**Modelo Operativo (Formalizable):**
$$x_{i,t+1} = x_{\text{best},t} + \alpha \cdot \xi \quad \text{(exploración local)}$$
$$x_{\text{best},t+1} = \arg\min\left(f(x_{i,t+1}) \text{ para } i=1..n\right)$$

Donde:
- $x_{\text{best}}$ = posición del "depredador" (mejor solución actual)
- $x_i$ = posiciones de "rémoras" (agentes explotadores)
- $\alpha$ = radio de exploración (adhesión virtual)
- $\xi$ = ruido aleatorio (perturbación local)
- $f()$ = función objetivo

**Ejemplo Concreto: Ajuste de Hiperparámetros ML**
- Espacio de búsqueda: learning_rate ∈ [0.0001, 0.1], num_capas ∈ [1, 20], regularización ∈ [0, 1]
- Depredadores iniciales: 3 configuraciones muy distintas (exploración global)
- Rémoras: 7 configuraciones cercanas a cada depredador (exploración local)
- Métrica: accuracy en validación
- Resultado: 10 agentes convergen → mejores hiperparámetros encontrados

### Heurística vs Metaheurística

**Heurística (Específica del Problema):**
"Para encontrar mejores hiperparámetros, incrementa el learning rate en pasos pequeños hasta que la validación empeore, luego retrocede."
- Estrategia ad-hoc
- Funciona solo para este parámetro
- No generalizable

**Metaheurística (Principio General):**
"Mantén múltiples puntos alrededor de tus mejores soluciones conocidas. Perturba cada uno localmente. Actualiza el mejor. Repite."
- Estrategia de propósito general
- Funciona para:
  - Hiperparámetros ML
  - Diseño de ingeniería (pesos, dimensiones)
  - Planificación de ruta (con distancia como función objetivo)
  - Predicción de estructura proteica
- Parámetro ajustable: radio de perturbación ($\alpha$)

**Pseudocódigo (Algoritmo Rémora):**
```
Inicio:
  depredadores ← generar_n_soluciones_aleatorias(n_exploradores)
  mejor_global ← evaluar(depredadores)
  
Para cada iteración t:
  rémoras ← generar_perturbaciones(mejor_global, radio=α)
  evaluaciones ← evaluar(rémoras)
  mejor_local ← argmin(evaluaciones)
  
  si mejor_local < mejor_global:
    mejor_global ← mejor_local
    depredadores ← actualizar_población(mejor_local, rémoras)
  
  si convergencia O iteraciones_max:
    Break
    
Retornar: mejor_global
```

### Comparación con Gradient Descent

| Aspecto | Gradient Descent | Algoritmo Rémora |
|--------|-----------------|------------------|
| **Tipo de Gradiente** | Calcula derivadas (∇f) | Sin derivadas (derivative-free) |
| **Exploración** | Siempre descendente (greedy) | Local pero con libertad direccional |
| **Paralelización** | Secuencial | Naturalmente paralelo (múltiples agentes) |
| **Requerimientos Matemáticos** | Función diferenciable | Solo necesita evaluación f(x) |
| **Óptimos Locales** | Queda atrapado fácilmente | Múltiples agentes = escape parcial |
| **Convergencia** | Rápida (cuadrática) si f es convexa | Lenta pero más robusta en no-convexa |
| **Aplicabilidad** | Modelos diferenciables (NN, regresión) | Problemas discretos, ruidosos, complejos |

**Similitudes:**
1. Ambos son algoritmos iterativos de optimización
2. Ambos buscan minimizar/maximizar una función objetivo
3. Ambos explotan "vecindad local" (cercanos al mejor punto)
4. Ambos pueden implementarse con épocas/iteraciones

**Diferencias:**

1. **Información Utilizada:**
   - GD: usa ∇f(x) = dirección de máximo descenso
   - Rémora: solo usa f(x), no requiere derivadas
   
2. **Exploración vs Explotación:**
   - GD: explotación pura (siempre sigue gradiente)
   - Rémora: balance (varios agentes = múltiples direcciones)
   
3. **Aplicación:**
   - GD: espacios continuos, funciones suaves
   - Rémora: espacios discretos, multimodales, ruidosos
   
4. **Robustez:**
   - GD: sensible a inicialización
   - Rémora: múltiples depredadores = menor sensibilidad
